```python
df['balcony']=df['balcony'].transform(lambda x:x.fillna(x.median()))
```

```python
# Outlier in bath i.e no of bathrooms
Q3=df_new['bath'].quantile(0.75)
Q1=df_new['bath'].quantile(0.25)

IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

#capping
df_new['bath']=df_new['bath'].clip(lower=lower_bound,upper=upper_bound)
```

```python
# 1. Clean extra spaces
df_new['location'] = df_new['location'].str.strip()

# 2. Get value counts
location_stats = df_new['location'].value_counts()

# 3. Identify locations with less than 10 entries
locations_less_than_10 = location_stats[location_stats < 10].index

# 4. Replace rare locations with 'other'
df_new['location'] = df_new['location'].apply(lambda x: 'other' if x in locations_less_than_10 else x)

# Check unique counts again
print("Remaining unique locations:", df_new['location'].nunique())
```

```python
df_new.head()
```

```python
# Drop 'society' due to high cardinality/missingness
df_new=df_new.drop(columns=['society'])
```

```python
# one-hot encoding
df_encoded = pd.get_dummies(df_new, columns=['area_type', 'location'], drop_first=True, dtype=int)
```

```python
df_encoded.head()
```

```python
df_encoded.info()
```

```python
df_encoded['total_sqft'].unique().tolist()
```

```python
# total_sqft column contains some string values in the form of range
def convert_sqft(x):
    if '-' in x:
        lower, upper = x.split('-')
        return (float(lower) + float(upper)) / 2

    if 'Sq. Meter' in x:
        return float(x.replace('Sq. Meter', '')) * 10.7639

    if 'Sq. Yards' in x:
      return float(x.replace('Sq. Yards','')) * 9

    if 'Acres' in x:
      return float(x.replace('Acres','')) * 43560

    return float(x)
```

```python
df_encoded['total_sqft']=df_encoded['total_sqft'].apply(convert_sqft)
```

# comparing regression metrics

```python
def compare_metrics(model_name,y_test,y_pred):
  mae=mean_absolute_error(y_test,y_pred)
  mse=mean_squared_error(y_test,y_pred)
  rmse=root_mean_squared_error(y_test,y_pred)
  r2=r2_score(y_test,y_pred)

  print(f"Metrics for {model_name}:")
  print(f"MAE: {mae}")
  print(f"MSE: {mse}")
  print(f"RMSE: {rmse}")
  print(f"R2: {r2}")
```

```python
compare_metrics('SVR',Y_test,svr_pred)
```

```python
compare_metrics('Decision Tree',Y_test,dt_pred)
```

```python
compare_metrics('Random Forest',Y_test,rf_pred)
```

```python
compare_metrics('Gradient Boosting',Y_test,gb_pred)
```

```python
compare_metrics('AdaBoost',Y_test,ada_pred)
```

```python
compare_metrics('XGBoost',Y_test,xgr_pred)
```

**Among all model Gradient boosting performs well with highest R^2 and lowest RMSE**

# Hyperparameter Tuning of Gradient Boosting Model

```python
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV
```

GridSearchCV

```python
gbr = GradientBoostingRegressor(random_state=42)

param_grid = {
    'n_estimators': [50, 100, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'max_depth': [2, 3, 4, 5]
}

grid_search = GridSearchCV(
    estimator=gbr,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1
)

grid_search.fit(X_train, Y_train)
```

```python
print(grid_search.best_params_)
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
```

```python
Y_pred = best_model.predict(X_test)
```

```python
r2 = r2_score(Y_test, Y_pred)
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)

print("R²:", r2)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```

RandomSearchCV

```python
gbr = GradientBoostingRegressor(random_state=42)

param_dist = {
    'n_estimators': [50, 100, 150, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1, 0.15, 0.2],
    'max_depth': [2, 3, 4, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4, 8],
    'subsample': [0.7, 0.8, 0.9, 1.0]
}

random_search = RandomizedSearchCV(
    estimator=gbr,
    param_distributions=param_dist,
    n_iter=30,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, Y_train)
```

```python
print("Best Parameters:")
print(random_search.best_params_)

print("Best CV R²:")
print(random_search.best_score_)
```

```python
best_model = random_search.best_estimator_
```

```python
Y_pred = best_model.predict(X_test)

print("Test R²:", r2_score(Y_test, Y_pred))
print("MAE:", mean_absolute_error(Y_test, Y_pred))
print("MSE:", mean_squared_error(Y_test, Y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, Y_pred)))
```

```python
best_random_model = random_search.best_estimator_

train_pred = best_random_model.predict(X_train)
test_pred = best_random_model.predict(X_test)

print("Train R²:", r2_score(Y_train, train_pred))
print("Test R² :", r2_score(Y_test, test_pred))
```

# Conclusion
Gradient Boosting Regressor achieved the highest cross-validation R² among the evaluated models. Hyperparameter tuning using GridSearchCV and RandomizedSearchCV was performed, but the tuned models did not improve the performance on the held-out test set. The original Gradient Boosting model therefore provided the best test-set performance among the tested configurations.
